Le rôle de chaque fichier est le suivant:
Fichier/répertoire 	Description
.cargo/config.toml 	Paramètres de compilation croisée pour ESP-IDF (target, linker, runner, etc.)
rust-toolchain.toml 	Verrouille la version Rust et la cible (esp)
sdkconfig.defaults 	Paramètres ESP-IDF (par exemple, options FreeRTOS)
build.rs 	Rust build script pour esp-idf-sys intégration
src/main.rs 	Le code d'application Rust écrit par l'utilisateur
Cargo.toml 	Définitions des dépendances et des caractéristiques de la Ruste
